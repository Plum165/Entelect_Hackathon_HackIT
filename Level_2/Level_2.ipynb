# 2. Processing Data

The processing stage is responsible for finding the optimal route through the Level 2 graph.

Level 2 is more complicated than Level 1 because we have four required stations:

- `S1`
- `S2`
- `S3`
- `S4`

All four stations must be visited before reaching `B`, but their order is not predetermined.

There are therefore two parts to the problem.

First, we need to find the shortest path between two individual nodes. We will use **Dijkstra's algorithm** for this.

Second, we need to determine the best order in which to visit the four required stations. Since there are only four stations, we can generate every possible ordering and evaluate each one.

The number of possible orderings is:

$$
4! = 4 \times 3 \times 2 \times 1 = 24
$$

Therefore, our algorithm will evaluate all 24 possible station orders and select the one with the lowest total cost.

For Level 2, the cost of an edge is:

$$
\text{effective cost} = \text{time} + \text{risk}
$$

The complete process is therefore:

**Input graph**

↓

**Calculate effective edge costs**

↓

**Dijkstra's algorithm**

↓

**Generate 24 station orders**

↓

**Find the shortest path for each leg**

↓

**Calculate total route cost**

↓

**Select the cheapest route**

`````python
import heapq
import itertools
`````

## 2.2 Calculating the Effective Edge Cost

In Level 1, each edge had a single `weight` representing travel time.

Level 2 contains two values for every edge:

- `time`
- `risk`

The challenge specifies that these should be combined to create the effective edge cost.

Therefore:

$$
\text{effective cost} = \text{time} + \text{risk}
$$

For example, consider the edge from `A` to `P6`:

```text
time = 5
risk = 2
Its effective cost is:

5+2=7
```

Dijkstra's algorithm will use this value when comparing possible routes.

`````python
def edge_cost(edge):
    return edge["time"] + edge["risk"]
`````

`````python
example_edge = level2_graph["A"][1]

print("Edge:", example_edge)
print("Effective cost:", edge_cost(example_edge))
`````

## 2.3 Dijkstra's Shortest Path Algorithm

We now implement Dijkstra's algorithm.

Dijkstra's algorithm finds the minimum-cost path between a starting node and a destination in a graph with non-negative edge costs.

The algorithm maintains two important pieces of information.

### Distance

The `distances` dictionary stores the cheapest known cost from the starting node to every other node.

Initially, the starting node has a cost of `0`, while every other node has a cost of infinity.

### Previous

The `previous` dictionary records which node was used to reach each node using the cheapest known route.

This allows us to reconstruct the actual path after Dijkstra has finished.

For Level 2, the algorithm does not use `time` by itself.

Instead, whenever an edge is examined, we calculate:

$$
\text{cost} = \text{time} + \text{risk}
$$

The node with the lowest current distance is processed next using a priority queue.

Once the destination is removed from the priority queue, its shortest distance has been determined.

`````python
def dijkstra(graph, start, end):

    # Store the cheapest known distance to each node.
    distances = {
        node: float("inf")
        for node in graph
    }

    # Store the previous node used to reach each node.
    previous = {
        node: None
        for node in graph
    }

    # The cost of reaching the start node is zero.
    distances[start] = 0

    # Priority queue stores:
    # (current distance, node)
    priority_queue = [(0, start)]

    while priority_queue:

        current_distance, current_node = heapq.heappop(
            priority_queue
        )

        # Ignore outdated entries.
        if current_distance > distances[current_node]:
            continue

        # The shortest path to the destination has been found.
        if current_node == end:
            break

        # Examine all neighbouring nodes.
        for neighbour_info in graph[current_node]:

            neighbour = neighbour_info["node"]

            # Level 2 cost = time + risk.
            cost = edge_cost(neighbour_info)

            new_distance = current_distance + cost

            # If this route is cheaper, update it.
            if new_distance < distances[neighbour]:

                distances[neighbour] = new_distance

                previous[neighbour] = current_node

                heapq.heappush(
                    priority_queue,
                    (new_distance, neighbour)
                )

    return distances, previous
`````

## 2.4 Reconstructing the Shortest Path

Dijkstra's algorithm gives us the shortest distance, but the challenge requires the actual sequence of nodes.

The `previous` dictionary allows us to reconstruct this sequence.

We start at the destination and repeatedly move to the node stored in `previous`.

For example, suppose Dijkstra produced:

```text
previous[B] = P5
previous[P5] = S4
previous[S4] = P4
previous[P4] = S1
```

We can trace the route backwards:

```
B → P5 → S4 → P4 → S1
```

We then reverse the result:

```
S1 → P4 → S4 → P5 → B
```

`````python
def reconstruct_path(previous, start, end):

    path = []

    current_node = end

    while current_node is not None:

        path.append(current_node)

        if current_node == start:
            break

        current_node = previous[current_node]

    path.reverse()

    return path
`````

## 2.5 Finding a Shortest Path Between Two Nodes

We can now combine Dijkstra's algorithm and path reconstruction.

The `shortest_path()` function accepts a starting node and an ending node.

It returns:

1. The shortest path between the two nodes.
2. The effective cost of that path.

This allows us to solve individual sections of the larger journey.

For example, if we need to travel from `A` to `S3`, we can call:

```text
shortest_path(graph, A, S3)
```

`````python
def shortest_path(graph, start, end):

    distances, previous = dijkstra(
        graph,
        start,
        end
    )

    path = reconstruct_path(
        previous,
        start,
        end
    )

    return path, distances[end]
`````

## 2.6 Generating All Possible Station Orders

The four required stations must all be visited, but the challenge does not specify their order.

We therefore need to consider every possible ordering.

The `itertools.permutations()` function generates every possible arrangement of the stations.

For four stations:

$$
4! = 24
$$

Therefore, there are exactly 24 possible station orders.

For example:

```text
S1 → S2 → S3 → S4

`````python
route = []

current_node = end_node

while current_node is not None:

    route.append(current_node)

    if current_node == start_node:
        break

    current_node = previous[current_node]

route.reverse()

print("Route:", route)
print("Cost:", distances[end_node])
`````

`````python
station_orders = list(
    itertools.permutations(required_stations)
)

print("Number of possible station orders:", len(station_orders))
`````

## 2.7 Evaluating a Station Order

A station order only contains the four required stations.

For example:

```text
S3 → S1 → S4 → S2
````

`````python
def evaluate_order(graph, start, end, station_order):

    # Create the complete sequence of checkpoints.
    checkpoints = (
        [start]
        + list(station_order)
        + [end]
    )

    total_cost = 0
    full_path = []

    # Process every consecutive pair of checkpoints.
    for i in range(len(checkpoints) - 1):

        leg_start = checkpoints[i]
        leg_end = checkpoints[i + 1]

        path, cost = shortest_path(
            graph,
            leg_start,
            leg_end
        )

        total_cost += cost

        # Add the complete first path.
        if i == 0:
            full_path.extend(path)

        # For later paths, skip the first node
        # because it is already in full_path.
        else:
            full_path.extend(path[1:])

    return full_path, total_cost
`````

## 2.8 Finding the Optimal Station Order

We can now solve the complete optimisation problem.

For every one of the 24 possible station orders, we:

1. Construct the complete journey from `A` to `B`.
2. Find the shortest path between every consecutive pair of checkpoints.
3. Calculate the total effective cost.
4. Compare the cost against the best route found so far.

The route with the smallest total cost is stored as our optimal solution.

This approach guarantees that we consider every possible ordering of the required stations.

Because there are only 24 combinations, checking every ordering is practical and ensures that we do not accidentally choose a locally good but globally poor ordering.

`````python
best_order = None
best_path = None
best_cost = float("inf")

for order in station_orders:

    path, cost = evaluate_order(
        level2_graph,
        start_node,
        end_node,
        order
    )

    if cost < best_cost:

        best_cost = cost
        best_order = order
        best_path = path
`````

## 2.9 Inspecting the Optimal Solution

After evaluating all 24 possible station orders, the algorithm has selected the route with the lowest effective cost.

We now inspect three values:

- `best_order` — the optimal order of the required stations.
- `best_path` — the complete route from `A` to `B`.
- `best_cost` — the total risk-adjusted cost.

The cost is based on:

$$
\text{total cost}
=
\sum(\text{time} + \text{risk})
$$

The cost is useful for verifying our algorithm, but it will not be included in the final submission file.

`````python
print("Optimal station order:")
print(best_order)

print()

print("Complete route:")
print(best_path)

print()

print("Total effective cost:")
print(best_cost)
`````
